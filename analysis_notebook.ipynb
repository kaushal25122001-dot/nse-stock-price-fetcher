# Task 2: Automating Price Data (NSE Stocks)

**Objective**: Download the past 30 days of daily closing prices for four selected NSE-listed stocks:
- HDFC Bank (`HDFCBANK.NS`)
- Reliance Industries (`RELIANCE.NS`)
- Tata Consultancy Services (`TCS.NS`)
- ITC Limited (`ITC.NS`)

### Step-by-Step Logic:
1. **Dynamic Date Math**: Use `datetime.now()` and `timedelta(days=30)` to calculate the start date dynamically without hardcoding.
2. **Adjusted Price Extraction**: Use `yfinance.download(..., auto_adjust=True)` to accurately incorporate stock splits and dividends.
3. **Data Sanitization**: Extract the `Close` attribute, rename ticker symbols to human-readable names, and format dates.
4. **Persistence**: Export the dataset to both CSV and Excel formats.

```python
# Step 1: Import required libraries
from datetime import datetime, timedelta
import pandas as pd
import yfinance as yf

# Define stock ticker mapping
TICKER_MAP = {
    'HDFCBANK.NS': 'HDFC Bank',
    'RELIANCE.NS': 'Reliance Industries',
    'TCS.NS': 'TCS',
    'ITC.NS': 'ITC'
}
```

```python
# Step 2: Compute dynamic 30-day date range
today = datetime.now().date()
start_date = (today - timedelta(days=30)).strftime('%Y-%m-%d')
# Note: yfinance end date is non-inclusive; add 1 day to capture today's session
end_date = (today + timedelta(days=1)).strftime('%Y-%m-%d')

print(f"Fetching prices from {start_date} to {today}")
```

```python
# Step 3: Fetch historical stock data
raw_data = yf.download(
    tickers=list(TICKER_MAP.keys()),
    start=start_date,
    end=end_date,
    auto_adjust=True
)

# Extract 'Close' prices (corporate action adjusted)
closing_df = raw_data['Close'].copy()

# Remap ticker symbols to company names
closing_df = closing_df.rename(columns=TICKER_MAP)
closing_df.index = closing_df.index.strftime('%Y-%m-%d')
closing_df.index.name = 'Date'
closing_df = closing_df.round(2)

# Display results
closing_df.tail(10)
```

```python
# Step 4: Export to CSV and Excel
closing_df.to_csv('past_30_days_closing_prices.csv')
try:
    closing_df.to_excel('past_30_days_closing_prices.xlsx')
    print("Exported to past_30_days_closing_prices.xlsx")
except Exception:
    pass

print("Exported to past_30_days_closing_prices.csv")
```
